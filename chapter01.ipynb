## 1.1 小节标题（示例：线性方程组与增广矩阵）

### 概念

$n$ 元线性方程组都可以写成矩阵形式 $A\mathbf{x}=\mathbf{b}$；把系数矩阵 $A$ 与右端项 $\mathbf{b}$ 左右拼在一起，就得到**增广矩阵** $[A \mid \mathbf{b}]$：

$$
A=\begin{bmatrix} 1 & 2 & 1 \\ 0 & 1 & -1 \\ 2 & 5 & 1 \end{bmatrix},\qquad
\mathbf{b}=\begin{bmatrix} 3 \\ 1 \\ 7 \end{bmatrix}
$$

### 推导：初等行变换保持解集不变

对增广矩阵做三种**初等行变换**（交换两行；某一行乘以非零常数；某一行的倍数加到另一行），不会改变方程组的解集。把 $[A \mid \mathbf{b}]$ 化为**行最简形**（RREF），主元列对应基本变量、非主元列对应自由变量，就能写出通解。

### Python 验证

下面用 SymPy 对增广矩阵做行化简，验证手算结果。

```python
# 目标: 用 SymPy 验证上面方程组的行化简与通解
import sympy as sp

# 1. 构造增广矩阵 [A | b]
A = sp.Matrix([
    [1, 2, 1],
    [0, 1, -1],
    [2, 5, 1],
])
b = sp.Matrix([3, 1, 7])
Ab = A.row_join(b)   # row_join: 把 b 拼到 A 的右边
print('增广矩阵 [A|b] =')
sp.pprint(Ab)

# 2. rref() 返回 (行最简形, 主元列下标)
R, pivots = Ab.rref()
print('\n行最简形 =')
sp.pprint(R)
print('主元所在列(从0计数):', pivots)

# 3. 通解: 令自由变量 x3 = t
#    由 R 的第 2 行: x2 - x3 = 1,  第 1 行: x1 + 2*x2 + x3 = 3
#    解得 x1 = 1 - 3t, x2 = 1 + t, x3 = t
#    任取 t 代入验证 A x = b 恒成立
t = sp.symbols('t')
x = sp.Matrix([1 - 3*t, 1 + t, t])
print('\n验证 A*x - b =')
sp.pprint((A * x - b).applyfunc(sp.simplify))   # 结果应为零向量
```
