## 1.1 Systems of Linear Equations

线性方程，形如
$$a_1x_1 + a_2x_2 + \dots + a_nx_n = b$$
$a$和$b$为实数或者复数。

多个包含相同变量的线性方程的组合被称为**线性方程组**，所有可能解的集合被称为**解集**。
若两个线性方程组有**相同的解集**，称这两个线性方程组**等价**。

$n$ 元线性方程组都可以写成矩阵形式 $A\mathbf{x}=\mathbf{b}$；把系数矩阵 $A$ 与右端项 $\mathbf{b}$ 左右拼在一起，就得到**增广矩阵** $[A \mid \mathbf{b}]$：
例如方程组:
$$
\begin{cases}
x_1 - 2x_2 + x_3 = 0 \\
2x_2 - 8x_3 = 8 \\
5x_1 - 5x_3 = 10
\end{cases}
$$
$$
A=\begin{bmatrix} 1 & -2 & 1 \\ 2 & 0 & -8 \\ 5 & 0 & -5 \end{bmatrix},\qquad
\mathbf{b}=\begin{bmatrix} 0 \\ 8 \\ 10 \end{bmatrix}
$$
$$
 [A \mid \mathbf{b}]=\begin{bmatrix} 1 & -2 & 1 & 0 \\ 2 & 0 & -8 & 8 \\ 5 & 0 & -5 & 10 \end{bmatrix}
$$

### 初等行变换
1. 交换两行；
2. 某一行乘以非零常数；
3. 某一行的倍数加到另一行

对**增广矩阵**做三种初等行变换不会改变方程组的解集。

### 线性方程组解的情况：
- 无解
- 唯一解
- 无穷解

有解consistent；无解inconsistent

### 几何意义--线性方程组的解
n个变量对应n维空间交点

例如2阶线性方程组求解问题--直线相交问题;3阶则对应三个平面求交点；n阶则对应n维

## 模板使用说明

1. 单元格分两类：**Markdown 单元格**写文字和公式，**代码单元格**写 Python。
2. 新建单元格：按 `Esc` 进入命令模式（单元格左边框变蓝），按 `a` 在上方插入、`b` 在下方插入；`m` 切换为 Markdown、`y` 切换为代码。
3. 运行单元格：`Shift + Enter`（运行并跳到下一格），`Ctrl + Enter`（只运行当前格）。
4. 数学公式：行内公式用 `$...$`，独立成行用 `$$...$$`，语法是 LaTeX。
5. 保存：`Ctrl + S`。做完一章后按流程 `git add → commit → push`。
