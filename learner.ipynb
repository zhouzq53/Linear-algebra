# 公式

$\mathbb{R}^n$

$\mathbf{x}$ 加粗向量


**列向量：**

$$
\mathbf{x}=
\begin{bmatrix}
x_1\\
x_2\\
x_3
\end{bmatrix}
$$

**矩阵：**

$$
A=
\begin{bmatrix}
1&2&3\\
4&5&6
\end{bmatrix}
$$

**增广矩阵：**

$$
[A\mid\mathbf{b}]=
\left[
\begin{array}{cc|c}
1&2&5\\
3&4&6
\end{array}
\right]
$$



**方程组：**

$$
\begin{cases}
2x+y=5\\
x-y=1
\end{cases}
$$

**多行推导：**

$$
\begin{aligned}
T(\mathbf{x})
&=T(x_1\mathbf{e}_1+x_2\mathbf{e}_2)\\
&=x_1T(\mathbf{e}_1)+x_2T(\mathbf{e}_2)\\
&=A\mathbf{x}
\end{aligned}
$$



**向量线性组合：**

$$
\mathbf{b}=c_1\mathbf{v}_1+c_2\mathbf{v}_2+\cdots+c_p\mathbf{v}_p
$$

**矩阵方程：**

$$
A\mathbf{x}=\mathbf{b}
$$

**齐次方程通解：**

$$
\mathbf{x}=s\mathbf{v}_1+t\mathbf{v}_2
$$

**非齐次方程通解：**

$$
\mathbf{x}=\mathbf{x}_p+\mathbf{x}_h
$$

**线性变换：**

$$
T:\mathbb{R}^n\to\mathbb{R}^m
$$

**线性性质：**

$$
T(c\mathbf{u}+d\mathbf{v})
=cT(\mathbf{u})+dT(\mathbf{v})
$$

**标准矩阵：**

$$
A=[T(\mathbf{e}_1)\quad T(\mathbf{e}_2)\quad\cdots\quad T(\mathbf{e}_n)]
$$

**反复迭代：**

$$
\mathbf{x}_{k+1}=M\mathbf{x}_k
$$
