$\mathbb{R}^n$

**列向量：**

$$
\mathbf{x}=
\begin{bmatrix}
x_1\\
x_2\\
x_3
\end{bmatrix}
$$

$$
A=
\begin{bmatrix}

$$

$$
[A\mid\mathbf{b}]=
\left[
\begin{array}{cc|c}
1&2&5\\
3&4&6

\right]
$$

$$
\begin{cases}
2x+y=5\\
x-y=1

**多行推导：**

$$
\begin{aligned}
T(\mathbf{x})
&=T(x_1\mathbf{e}_1+x_2\mathbf{e}_2)\\
&=x_1T(\mathbf{e}_1)+x_2T(\mathbf{e}_2)\\
&=A\mathbf{x}
\end{aligned}
$$


## 第一章常用公式

**向量线性组合：**

$$
\mathbf{b}=c_1\mathbf{v}_1+c_2\mathbf{v}_2+\cdots+c_p\mathbf{v}_p
$$

**矩阵方程：**

$$
A\mathbf{x}=\mathbf{b}
$$

**齐次方程通解：**

$$
\mathbf{x}=s\mathbf{v}_1+t\mathbf{v}_2
$$

**非齐次方程通解：**

$$
\mathbf{x}=\mathbf{x}_p+\mathbf{x}_h
$$

**线性变换：**

$$
T:\mathbb{R}^n\to\mathbb{R}^m
$$

**线性性质：**

$$
T(c\mathbf{u}+d\mathbf{v})
=cT(\mathbf{u})+dT(\mathbf{v})
$$

**标准矩阵：**

$$
A=[T(\mathbf{e}_1)\quad T(\mathbf{e}_2)\quad\cdots\quad T(\mathbf{e}_n)]
$$

**反复迭代：**

$$
\mathbf{x}_{k+1}=M\mathbf{x}_k
$$


# Python 矩阵运算常用函数

```python
##代码范例

import numpy as np
import sympy as sp

# 创建矩阵和右端向量
A = sp.Matrix([
    [2, 1],
    [1, -1]
])
b = sp.Matrix([5, 1])

# 1. 构造增广矩阵
Aug = A.row_join(b)
print("增广矩阵：")
sp.pprint(Aug)

# 2. 行最简形与主元位置
R, pivots = Aug.rref()
print("\n行最简形：")
sp.pprint(R)
print("主元列（从 1 开始）：", [i + 1 for i in pivots])

# 3. 求解方程组
print("\n方程组解集：")
print(sp.linsolve((A, b)))

# 4. 数值计算
A_np = np.array(A.tolist(), dtype=float)
b_np = np.array(b, dtype=float).reshape(-1)
x = np.linalg.solve(A_np, b_np)

print("\n数值解：", x)
print("验证 Ax =", A_np @ x)
```
